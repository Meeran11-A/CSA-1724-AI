# 8-Queens Problem Using Backtracking


```python
N = 8
board = [-1] * N  # board[row] stores the column containing that row's queen

def is_safe(row, col):
    for previous_row in range(row):
        previous_col = board[previous_row]
        # Check same column or diagonal
        if previous_col == col or abs(previous_col - col) == abs(previous_row - row):
            return False
    return True

def solve_queens(row=0):
    if row == N:
        return True  # All eight queens have been placed

    for col in range(N):
        if is_safe(row, col):
            board[row] = col
            if solve_queens(row + 1):
                return True
            board[row] = -1  # Backtrack
    return False

def display_board():
    for row in range(N):
        print(' '.join('Q' if board[row] == col else '.' for col in range(N)))

if solve_queens():
    print('One solution to the 8-Queens problem:')
    display_board()
else:
    print('No solution exists.')
```
