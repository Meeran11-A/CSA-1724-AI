# Water Jug Problem Using Breadth-First Search (BFS)


```python
from collections import deque

def water_jug_problem(capacity1, capacity2, target):
    # Each state is (amount of water in jug 1, amount in jug 2)
    start = (0, 0)
    queue = deque([(start, [start])])
    visited = {start}

    while queue:
        (a, b), path = queue.popleft()

        if a == target or b == target:
            return path

        # Generate all possible actions
        next_states = [
            (capacity1, b),  # Fill jug 1
            (a, capacity2),  # Fill jug 2
            (0, b),          # Empty jug 1
            (a, 0),          # Empty jug 2
        ]

        # Pour jug 1 into jug 2
        pour = min(a, capacity2 - b)
        next_states.append((a - pour, b + pour))

        # Pour jug 2 into jug 1
        pour = min(b, capacity1 - a)
        next_states.append((a + pour, b - pour))

        for state in next_states:
            if state not in visited:
                visited.add(state)
                queue.append((state, path + [state]))

    return None

# Jug capacities and target amount (in litres)
jug1_capacity = 4
jug2_capacity = 3
target_amount = 2

solution = water_jug_problem(jug1_capacity, jug2_capacity, target_amount)

if solution:
    print('Steps to solve the Water Jug Problem:')
    for step, (jug1, jug2) in enumerate(solution):
        print(f'Step {step}: Jug 1 = {jug1} L, Jug 2 = {jug2} L')
    print('\nTarget achieved!')
else:
    print('No solution exists for the given capacities and target.')
```
