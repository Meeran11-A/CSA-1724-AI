```python
from collections import deque


def is_valid_state(m_left: int, c_left: int, total_m: int = 3, total_c: int = 3) -> bool:
    m_right = total_m - m_left
    c_right = total_c - c_left

    # Bounds check
    if not (0 <= m_left <= total_m and 0 <= c_left <= total_c):
        return False

    # Left bank safety check
    if m_left > 0 and m_left < c_left:
        return False

    # Right bank safety check
    if m_right > 0 and m_right < c_right:
        return False

    return True


def solve_missionaries_cannibals(total_m: int = 3, total_c: int = 3):
    """
    Finds the shortest sequence of states from (total_m, total_c, 1) to (0, 0, 0)
    using Breadth-First Search (BFS). Boat capacity is 1 or 2 people.
    """
    start = (total_m, total_c, 1)
    goal = (0, 0, 0)

    # All valid boat loads (missionaries, cannibals)
    moves = [(1, 0), (2, 0), (0, 1), (0, 2), (1, 1)]

    queue = deque([(start, [start])])
    visited = {start}

    while queue:
        (m_left, c_left, boat), path = queue.popleft()

        if (m_left, c_left, boat) == goal:
            return path

        for dm, dc in moves:
            if boat == 1:
                # Boat moves from Left -> Right
                next_state = (m_left - dm, c_left - dc, 0)
            else:
                # Boat moves from Right -> Left
                next_state = (m_left + dm, c_left + dc, 1)

            if is_valid_state(next_state[0], next_state[1], total_m, total_c) and next_state not in visited:
                visited.add(next_state)
                queue.append((next_state, path + [next_state]))

    return None


def print_solution(path):
    if not path:
        print("No solution found.")
        return

    print(f"Solved in {len(path) - 1} boat crossings:\n")
    for step, (m_l, c_l, boat) in enumerate(path):
        m_r, c_r = 3 - m_l, 3 - c_l
        side = "Left Bank" if boat == 1 else "Right Bank"

        if step == 0:
            action = "Start"
        else:
            prev_m_l, prev_c_l, prev_boat = path[step - 1]
            if prev_boat == 1:
                moved_m = prev_m_l - m_l
                moved_c = prev_c_l - c_l
                action = f"Move {moved_m}M, {moved_c}C -> Right"
            else:
                moved_m = m_l - prev_m_l
                moved_c = c_l - prev_c_l
                action = f"Move {moved_m}M, {moved_c}C <- Left"

        print(
            f"Step {step:2d} | {action:<22} | "
            f"Left: ({m_l}M, {c_l}C) | Right: ({m_r}M, {c_r}C) | Boat: {side}"
        )


if __name__ == "__main__":
    solution_path = solve_missionaries_cannibals()
    print_solution(solution_path)
```
