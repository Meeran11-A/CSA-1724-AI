```python
from collections import deque


def bfs_traversal(graph: dict, start):
    """
    Performs Breadth-First Search (BFS) traversal starting from `start`
    and returns the list of nodes in the order they were visited.
    """
    visited = {start}
    queue = deque([start])
    order = []

    while queue:
        node = queue.popleft()
        order.append(node)

        for neighbor in graph.get(node, []):
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

    return order


def bfs_shortest_path(graph: dict, start, goal):
    """
    Finds the shortest path (fewest edges) between `start` and `goal` using BFS.
    Returns the path as a list of nodes, or None if unreachable.
    """
    if start == goal:
        return [start]

    visited = {start}
    queue = deque([(start, [start])])

    while queue:
        node, path = queue.popleft()

        for neighbor in graph.get(node, []):
            if neighbor == goal:
                return path + [neighbor]
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append((neighbor, path + [neighbor]))

    return None


if __name__ == "__main__":
    # Sample unweighted graph (Adjacency List)
    graph = {
        "A": ["B", "C"],
        "B": ["A", "D", "E"],
        "C": ["A", "F"],
        "D": ["B"],
        "E": ["B", "F"],
        "F": ["C", "E"],
    }

    print("BFS Traversal starting from 'A':", bfs_traversal(graph, "A"))
    print("Shortest path from 'A' to 'F':", bfs_shortest_path(graph, "A", "F"))
```
