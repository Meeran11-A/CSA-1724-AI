```python
import heapq


def a_star_search(graph: dict, heuristics: dict, start: str, goal: str):
    """
    Finds the lowest-cost path from `start` to `goal` using the A* algorithm.

    Parameters:
        graph: dict mapping node -> list of (neighbor, edge_cost)
        heuristics: dict mapping node -> estimated cost h(node) to goal
        start: starting node
        goal: target node

    Returns:
        (total_cost, path) if a path exists, otherwise (float('inf'), None)
    """
    # Priority queue stores tuples: (f_score, g_score, current_node, path)
    open_heap = [(heuristics[start], 0, start, [start])]

    # Best known g_score (cost from start) for each visited node
    best_g = {start: 0}

    while open_heap:
        f_score, g_score, current, path = heapq.heappop(open_heap)

        if current == goal:
            return g_score, path

        # Skip outdated entries in the priority queue
        if g_score > best_g.get(current, float("inf")):
            continue

        for neighbor, weight in graph.get(current, []):
            tentative_g = g_score + weight
            if tentative_g < best_g.get(neighbor, float("inf")):
                best_g[neighbor] = tentative_g
                new_f = tentative_g + heuristics.get(neighbor, 0)
                heapq.heappush(open_heap, (new_f, tentative_g, neighbor, path + [neighbor]))

    return float("inf"), None


if __name__ == "__main__":
    # Weighted graph: node -> [(neighbor, cost), ...]
    graph = {
        "S": [("A", 1), ("G", 10)],
        "A": [("B", 2), ("C", 1)],
        "B": [("D", 5)],
        "C": [("D", 3), ("G", 4)],
        "D": [("G", 2)],
        "G": [],
    }

    # Admissible heuristic estimates h(n) to reach goal 'G'
    heuristics = {
        "S": 5,
        "A": 3,
        "B": 4,
        "C": 2,
        "D": 6,
        "G": 0,
    }

    cost, path = a_star_search(graph, heuristics, "S", "G")
    print(f"Optimal Path from 'S' to 'G': {' -> '.join(path)}")
    print(f"Total Path Cost: {cost}")
```
