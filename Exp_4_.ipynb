```python
def solve_cryptarithmetic(w1: str, w2: str, w3: str):
    """
    Solves a cryptarithmetic addition puzzle: w1 + w2 = w3
    where each unique letter maps to a distinct digit (0-9),
    and leading letters of multi-letter words cannot be 0.
    """
    w1, w2, w3 = w1.upper(), w2.upper(), w3.upper()

    # Collect unique letters in order of appearance
    letters = []
    for ch in w1 + w2 + w3:
        if ch.isalpha() and ch not in letters:
            letters.append(ch)

    if len(letters) > 10:
        raise ValueError("More than 10 unique letters; cannot map to digits 0-9.")

    leading_letters = set()
    if len(w1) > 1:
        leading_letters.add(w1[0])
    if len(w2) > 1:
        leading_letters.add(w2[0])
    if len(w3) > 1:
        leading_letters.add(w3[0])

    def word_to_num(word: str, mapping: dict) -> int:
        num = 0
        for ch in word:
            num = num * 10 + mapping[ch]
        return num

    used_digits = [False] * 10
    mapping = {}
    solutions = []

    def backtrack(idx: int):
        if idx == len(letters):
            n1 = word_to_num(w1, mapping)
            n2 = word_to_num(w2, mapping)
            n3 = word_to_num(w3, mapping)
            if n1 + n2 == n3:
                solutions.append((mapping.copy(), n1, n2, n3))
            return

        ch = letters[idx]
        start_digit = 1 if ch in leading_letters else 0

        for digit in range(start_digit, 10):
            if not used_digits[digit]:
                used_digits[digit] = True
                mapping[ch] = digit
                backtrack(idx + 1)
                del mapping[ch]
                used_digits[digit] = False

    backtrack(0)
    return solutions


def display_solutions(w1: str, w2: str, w3: str):
    w1, w2, w3 = w1.upper(), w2.upper(), w3.upper()
    sols = solve_cryptarithmetic(w1, w2, w3)

    if not sols:
        print(f"No solution found for {w1} + {w2} = {w3}")
        return

    print(f"Found {len(sols)} solution(s) for {w1} + {w2} = {w3}:\n")
    for i, (mapping, n1, n2, n3) in enumerate(sols, start=1):
        print(f"--- Solution {i} ---")
        print("Mapping:", ", ".join(f"{k}={v}" for k, v in mapping.items()))
        print(f"  {n1}")
        print(f"+ {n2}")
        print("-" * (max(len(str(n1)), len(str(n2)), len(str(n3))) + 2))
        print(f"  {n3}\n")


# Example 1: Classic SEND + MORE = MONEY
display_solutions("SEND", "MORE", "MONEY")

# Example 2: Another classic puzzle
# display_solutions("TWO", "TWO", "FOUR")
```
