```python

from itertools import permutations

def travelling_salesman(graph, start):
    n = len(graph)
    cities = list(range(n))
    cities.remove(start)

    min_cost = float('inf')
    best_path = None

    # Generate all possible routes
    for route in permutations(cities):
        path = [start] + list(route) + [start]

        # Calculate total distance
        cost = 0
        for i in range(len(path) - 1):
            cost += graph[path[i]][path[i + 1]]

        # Find minimum cost route
        if cost < min_cost:
            min_cost = cost
            best_path = path

    return best_path, min_cost


# Distance matrix
graph = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]

start_city = 0

path, cost = travelling_salesman(graph, start_city)

city_names = ['A', 'B', 'C', 'D']

print("Travelling Salesman Problem")
print("Optimal Path:", " -> ".join(city_names[i] for i in path))
print("Minimum Travel Cost:", cost)
```
